Pandas Indexing Refresher

```python
import pandas as pd

df=pd.DataFrame({
                "name": ["Oak","Pine","Maple", "Cedar"],
                "ft":[25,48,31,60],
                "zone":["A","B","A","B"]
})
```

```python
df["ft"]
```

```python
df[["name","ft"]]
```

```python
df.iloc[1]
```

```python
df.iloc[0:3]
```

```python
df[df["ft"] > 30]
```

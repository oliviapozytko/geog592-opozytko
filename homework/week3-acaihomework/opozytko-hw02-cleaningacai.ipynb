1. Inspect the original CSV

With this step, i told python to read the csv, take al the rows from the csv and put them into a python list. Then i asked it to print the first 10 rows and print each one. 

```python
import csv

filename = "tabela1613.csv"

with open(filename, "r", encoding="utf-8-sig") as file:
    reader = csv.reader(file)

    rows = list(reader)

for row in rows[:10]:
    print(row)
```

2. Inspect the data and identify the columns needed for this assignment.
 In this step, I am filtering fata that only has municiplalities in Para. I created an empty list where i stored only Para records with a municplality column. I then checked the first 10 rows to make sure the what I did worked.

```python
para_rows = []

for row in rows:
    if len(row) > 2 and row[2].endswith("(PA)"):
        para_rows.append(row)

for row in para_rows[:10]:
    print(row)
```

3. Filter the data so that only municipalities in Pará (PA) remain.

I did this to keep the municipalities. I made a list for selected_rows, told python to go through each para municipality, go through each one to get only the two values being the cod row and the acai row

```python
selected_rows = []

for row in para_rows:
    code = row[1]
    acai = row[5]

    selected_rows.append([code, acai])

for row in selected_rows[:10]:
    print(row)
```

1. Inspect the original CSV

With this step, i told python to read the csv, take all the rows from the csv and put them into a python list. 

```python
import csv #import the csv

filename = "tabela1613.csv" # tell what file we are using 

with open(filename, "r", encoding="utf-8-sig") as file:    #open and read the original CSV
    reader = csv.reader(file)

    rows = list(reader) # store all rows in a list called rows

for row in rows[:10]:
    print(row) #print 10 to see if it is correct!
```

2. Inspect the data and identify the columns needed for this assignment.


 In this step, I am filtering data that only has municiplalities in Para. 

```python
para_rows = [] #make a list called para_rows

for row in rows:   #select the rows where the municipality ends in (PA) for the Para Municipality 

    if len(row) > 2 and row[2].endswith("(PA)"): #row 2 is the municpality column 
        para_rows.append(row)

for row in para_rows[:10]:
    print(row)
```

3. Filter the data so that only municipalities in Pará (PA) remain.
 

```python
selected_rows = [] 

for row in para_rows:   #keeping only the columns i need which is code and acai 
    code = row[1] #municipality code 
    acai = row[5] #acai value 

    selected_rows.append([code, acai])

for row in selected_rows[:10]:
    print(row)
```

4. Keep only the columns containing: the municipality code (Cod / Cód.), which will later be used to join the CSV to a GIS layer; the quantity of açaí produced. 

```python
for row in selected_rows: #i am looking to see if the values are blank 
    if row[0] == "" or row[1] == "":
        print("Missing value:", row)

for row in selected_rows: #i am looking for non-numeric acai values 
    if not row[1].isdigit():
        print("Non-numeric Açaí value:", row)

for row in selected_rows[:5]: #i am checking the current data types and what python considers its values (string, booelan, integer,etc. )
    print(row, type(row[0]), type(row[1]))
```

```python
# next i cleaned the rows to see the values and make sure theyre not blank or numeric 

cleaned_rows = []

for row in selected_rows:
    code = row[0]
    acai = row[1]

    if code != "" and acai.isdigit(): #checking for numeric and blank values 
        cleaned_rows.append([code, int(acai)])

for row in cleaned_rows[:10]:
    print(row) #checking the first 10 
```

```python
output_file = "opozytko-hw02-cleaned_para_acai_data.csv"

with open(output_file, "w", encoding="utf-8-sig", newline="") as file:
    writer = csv.writer(file)

    # Write the column names
    writer.writerow(["Cod", "Acai"])

    # Write all of the cleaned data
    writer.writerows(cleaned_rows)

print("Cleaned CSV created")
```

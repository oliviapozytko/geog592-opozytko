# Part 2: Find a Location Using a Loop

````python
# Your answer to Part 2: using code: 
range(n)

for index in range(n):                          #for each index in the range of 0,1,2,3(ABCD)
    if locations[index] == "C":                 #if the location at the index is equal to C
        print("Index:", index)                  #print the index
        print("Location:", locations[index])     #print the location of the index.....
        print("Easting:", eastings[index])
        print("Northing:", northings[index])         
````

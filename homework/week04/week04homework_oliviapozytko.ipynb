```python

cleaned_rows = []

with open(input_file, "r", encoding="utf-8-sig", newline="") as file:
    reader = csv.DictReader(file)

    for row in reader:
        if row["Technology"].strip() != "Solar PV":
            continue

        try:
            float(row["Installed_Capacity_MW"])
            float(row["Capacity_Factor_Percent"])
            float(row["Annual_Generation_MWh"])
            float(row["Capital_Cost_USD_per_kW"])
            float(row["Estimated_CO2_Avoided_tons_per_year"])
            int(row["Year"])
        except (ValueError, TypeError):
            continue

        cleaned_rows.append(row)
#check the data types to check if it is string, bool, or int, or decimal. 
```

```python
columns_to_keep = [
    "Project_ID",
    "Country",
    "Year",
    "Technology",
    "Installed_Capacity_MW",
    "Capacity_Factor_Percent",
    "Annual_Generation_MWh",
    "Estimated_CO2_Avoided_tons_per_year"
]
```

```python
final_rows = []

for row in cleaned_rows:
    new_row = {}

    for column in columns_to_keep:
        new_row[column] = row[column]

    final_rows.append(new_row)

    print("Number of cleaned rows:", len(final_rows))

for row in final_rows[:5]:
    print(row)
```

```python
with open(output_file, "w", encoding="utf-8", newline="") as file:
    writer = csv.DictWriter(file, fieldnames=columns_to_keep)

    writer.writeheader()
    writer.writerows(final_rows)

    # turn into a CSV
```

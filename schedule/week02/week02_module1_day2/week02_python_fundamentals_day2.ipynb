## 4. Comparison and Logical Operators

Conditions compare values.

### Comparison operators

| Operator | Meaning |
|---|---|
| `>` | greater than |
| `<` | less than |
| `>=` | greater than or equal to |
| `<=` | less than or equal to |
| `==` | equal to |
| `!=` | not equal to |

A comparison evaluates to either `True` or `False`.

### Combining conditions with `and` and `or`

Use **`and`** when **both conditions must be true**.

```python
distance = 750
distance > 500 and distance < 1000
```

Use **`or`** when **at least one condition must be true**.

```python
city = "Durham"
city == "Durham" or city == "Raleigh"
```

Remember:

- `and` → all required conditions
- `or` → at least one acceptable condition


````python
distance = 750

print(distance > 500)
print(distance < 500)
print(distance == 750)
````

## 5. `if` Statements

An `if` statement tells Python to run code **only when a condition is true**.

```python
distance = 750

if distance > 500:
    print("Long distance")
```

### Don't forget the `:` !

The colon separates the **condition** from the **code block** that is controlled by the condition.

A colon is required after statements such as:

```python
if condition:
else:
for item in collection:
```

Without the colon, Python raises a `SyntaxError`.

Incorrect:

```python
if distance > 500
    print("Long distance")
```

Correct:

```python
if distance > 500:
    print("Long distance")
```

###  Indentation is important!

Python uses **indentation** to define which lines belong to the same code block.

```python
if distance > 500:
    print("Long distance")

print("Finished")
```

Here:

- `print("Long distance")` belongs to the `if` block and runs only when the condition is true.
- `print("Finished")` is outside the block and runs every time.

Unlike languages that use braces `{ }` such as r and JS, Python uses indentation as part of its syntax.

You should use **4 spaces** for each indentation level (some editors and IDE will let you choose how to indent, but it is easiest to just be consistent.)

Incorrect:

```python
if distance > 500:
print("Long distance")
```

Correct:

```python
if distance > 500:
    print("Long distance")
```

Nested blocks use additional indentation:

```python
if distance > 500:
    if distance < 1000:
        print("Medium-long distance")
```


````python
distance = 750

if distance > 500:
    print("Long distance")
````

### Using `and` inside an `if`

Use `and` when both conditions must be satisfied.

````python
distance = 750

if distance > 500 and distance < 1000:
    print("Distance is between 500 and 1000 meters")
````

### Using `or` inside an `if`

Use `or` when either condition is acceptable.

````python
city = "Durham"

if city == "Durham" or city == "Raleigh":
    print("City is in the Triangle")
````

## 7. `for` Loops
A `for` loop repeats an action for each item in a collection.

Read this as: **for each city in cities, print that city.**


````python
cities = ["Chapel Hill", "Durham", "Raleigh"]

for city in cities:
    print(city)
````

The previous loop should have produced three print outputs since we had three elements in the list. 
We can count how many times a loop runs by using a **counter**.

A counter is a variable that we create **before** the loop. We usually start it at `0`, then increase it by `1` each time the loop runs.

For example:

````python
cities = ["Chapel Hill", "Durham", "Raleigh"]
counter = 0

for city in cities:
    counter = counter + 1
    print("Loop number:", counter)
    print("Location:", city)
````

Sometimes we want a loop to repeat a specific number of times.

Python's `range()` function generates a sequence of numbers that we can use in a loop.

For example:

````python
n = 8

for number in range(n):

    if number == 0:
        print(number, "notice that the range started at 0")

    elif number == (n - 1):
        print(number, "notice that the range finished at", n - 1)

    else:
        print(number)
````

Hey notice that I used another conditional. I used `elif` , which means “else if.”

It lets Python test another condition only if the previous `if` condition was false.

## 8. Combining a Loop and a Condition
This is a programming pattern that you're likely to use many times:

1. loop through a collection, then
2. test each value, and
3. execute a command *only* when the condition is true.


````python
distances = [300, 700, 1200, 450]

for distance in distances:
    if distance > 500:
        print(distance)
````

Expected output:

```text
700
1200
```

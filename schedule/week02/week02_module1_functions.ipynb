## 1. Why use functions?

Instead of repeating the same logic several times, we can write it once and reuse it.  This makes code easier to read and debug.

For example, converting from miles to kilometers

```python
distance1_km = distance1_miles * 1.60934
distance2_km = distance2_miles * 1.60934
distance3_km = distance3_miles * 1.60934
```

can be done in one reusable function.



## 2. Your first function

A function begins with `def`.

Look at the example below.  Notice that a declaration requires
- the function name
- parentheses `()`
- and a colon `:`

The body of the function is separated from the rest of your code using *indentation.*


````python
def say_hello():
    print("Welcome to GEOG 592")
````


Defining the function does **not** run it.

To run the function, we **call** it:


````python
say_hello()
````


### Anatomy of the function

```python
def say_hello():
    print("Welcome to GEOG 592")
```

- `def` tells Python that we are defining a function.
- `say_hello` is the function name.
- `()` contains parameters, if the function needs any.
- `:` marks the start of the function block.
- the indented code belongs to the function.



## 3. Parameters and arguments

A **parameter** is a name listed (inside the parentheses) in the function declaration. It behaves like a variable in the function body.


````python
def greet_student(name):
    print(f"Welcome, {name}")
````

 Arguments are the values or other variables provided to the function by putting them in parentheses when the function is called.  
 
 The corresponding parameter in the function body will be initially set the value of the argument.

````python
greet_student("Maria")
greet_student("Carlos")
````


In:

```python
def greet_student(name):
```

`name` is the **parameter**.

In:

```python
greet_student("Maria")
```

`"Maria"` is the **argument**.



## 4. Another example function

````python
miles_to_km(10)
miles_to_km(25)
````


This works, but the function only displays the answer.

What if we want to save the result and use it later?



## 5. `return`

`return` sends a value back to the place where the function was called. 

````python
def miles_to_km(miles):
    kilometers = miles * 1.60934
    return kilometers
````

````python
distance = miles_to_km(10) ## here, the parameter miles, of the function miles_to_km, is given the argument 10, 
# The function returns the argument converted to kilometers. 
# The returned value is assigned to the variable distance.  

print(distance)
````


### `print()` versus `return`

`print()` displays a result.

`return` gives the result back so another part of the program can use it.


````python
distance = miles_to_km(10)

double_distance = distance * 2

print(double_distance)
````

## `return` also ends the function

When Python reaches a `return` statement, the function stops immediately.

Any code inside the function that comes after `return` will not run.

This is similar to how `break` stops a loop, but there is an important difference:

- `return` stops the entire function and can send a value back. It only works inside functions.
- `break` stops a loop.

For example: 

````python
def check_value(value):

    if value < 0:
        return "Invalid value"

    print("The value is >= 0")

    return "Valid value"

### run the function with different values to see how it behaves
````

````python
for number in range(10):
    if number == 4:
        break
    print(number)
````


## 6. More than one parameter
Functions can have as many parameters as you wish. However you should consider that the more parameters you use, the more complicated and difficult it becomes to remember how to call the function. 

````python
def rectangle_area(width, height):
    area = width * height
    return area
````

````python
area = rectangle_area(20, 10)

print(area)
````


In:

## 7. Functions can make decisions


````python
def classify_distance(distance):

    if distance < 1000:
        return "near"
    else:
        return "far"
````

````python
print(classify_distance(450))
print(classify_distance(2200))
````


## 8. Returning `True` or `False`

Functions can answer yes/no questions with Boolean values.


````python
def is_positive(value):

    if value > 0:
        return True
    else:
        return False
````

````python
print(is_positive(25))
print(is_positive(-4))
````


## 9. Side quest: error handling

Later, when we work with CSV files, all values will initially be imported as strings.  We may want to know if the *value can successfully be converted to a number.*

One way to do this would be to check every character in the string and make sure each of them is numeric in it.  But you would have to also allow '-' and '.' to be characters to include negative numbers and floats. And there could only be one negative sign and one period, and the negative sign would need to be at the beginning of the string...

Python already has functions to convert strings to numbers, and will print an error message when you try to convert an invalid string:

````python
# Example of Python's built-in error checking

#this is fine
good_float=float('123.4')
print(good_float)
#this causes an error
bad_float = float('--123.4')
print('My only dream is to be printed when this cell executes.') #too bad...
````

The code above caused a `ValueError` error.

**Error handling** allows you to use Python's internal error checks without ending your code's execution. 

You instruct Python to `try` a block of (indented!) code.  If it causes an error matching the type listed in the `except` statement, it will execute that code.

````python
#this is fine
string_but_not_float = '--123.4'
try: 
    bad_float = float(string_but_not_float)
except ValueError:
    print(f'Oops, someone entered {string_but_not_float} in the CSV.') #the program never gets here
print('Life goes on.')    
````



An unqualified `except:` will catch *any* error.

That's not always a good thing! 

If you want Python to help you find errors in your code, *you should specifically list only the errors you expect* in the except statement.

For example, the following code runs, and prints unhelpful information for the programmer.

## 10. Using error handling in functions

Here's a function that uses error handling to *test* whether or not a value can be converted to a float.

````python
def can_be_float(value):

    try:
        float(value)
        return True

    except ValueError:
        return False
````

````python
print(can_be_float("24.5"))
print(can_be_float("30"))
print(can_be_float("hello"))
````


You do not need to master `try` / `except` today, but you should understand what the function `can_be_float(value)` does.

## 10. Functions inside loops

````python
values = ["23.4", "18.7", "error", "31.2"]

for value in values:

    if not can_be_float(value):
        print("Suspicious value:", value)
````

# Key ideas

A function often follows this pattern:

```python
def function_name(parameter):
    # code
    return result
```

Think of it as:

```text
INPUT
  ↓
FUNCTION
  ↓
OUTPUT
```

Functions help us:

- avoid repeating code;
- give a task a meaningful name;
- reuse the same logic;
- make programs easier to read;
- apply the same rule to many records.

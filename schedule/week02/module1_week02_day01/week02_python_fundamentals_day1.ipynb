## 2. Variables
A **variable** is a name that refers to a value.


- `=` assigns a value.
- `500` is the value stored in the variable.


````python
distance = 500
print(distance)
````

## 3. Basic Data Types

| Data Type | Meaning | Example |
|---|---|---|
| `int` | Whole number | `25` |
| `float` | Decimal number | `25.5` |
| `str` | Text | `"Chapel Hill"` |
| `bool` | True/False value | `True` |

Use the function `type()` to find the data type of a variable.  Basic data types (as well as more complicated object types) are considered *classes* in Python.


````python
city = "Chapel Hill"
population = 61960
area = 21.3
is_in_nc = True

print(type(city))
print(type(population))
print(type(area))
print(type(is_in_nc))
````

> Note that resolution has class 'float' even though you might consider 2.0 to be a whole number.  Python interprets the decimal point as an instruction to use a non-integer data type.

## 5. Strings
A Python string is text enclosed in quotation marks. You can use either double or single quotes (but not a combination of both).

````python
city = "Durham"
state = 'NC'
assert 'Hello' == "Hello"

print(city)
print(state)
````

### Concatenation  

Concatenation is joining two or more strings together to create one string.  In Python, you can use the `+` operator to concatenate.
> Don't confuse the word joining above with the GIS operation. 

````python
city = "Durham"
state = "NC"

location = city + ", " + state
print(location)
````

## 6. f-Strings
An **f-string** is an easy way to place variable values inside text.

Notice the `f` before the quote and the braces `{}` around variable names.


````python
place = "Davis Library"
distance = 350

print(f"{place} is {distance} meters away.")
````

## Working with GeoJSON Structures

### Today we will practice using useful data structures beyond lists:
We will learn about dictionaries (key/value pairs) and build nested data structures (dictionaries with lists as values, lists with dictionaries as values, ...).
We will also learn the hierarchies of the GeoJSON Feature and FeatureCollection types.  You will learn to retrieve data from nested structures and to loop through GeoJSON features.


Today we are **not** opening a `.geojson` file yet.

The purpose of this class is to understand how GeoJSON information is organized once it is represented in Python through dictionaries and lists.


## 1. From Module 1 Lists to Dictionaries

In Module 1, we used lists:

```python
locations = ["A", "B", "C", "D"]
```

A list stores values in order. We retrieve a value using its **index**.


````python
locations = ["A", "B", "C", "D"]

print(locations[0])
print(locations[2])
````

### A dictionary is defined with curly braces `{}` and stores information using *keys.*

*Keys* can be data of many types, including strings, integers, and floats (but not lists!)

```python
locationDict = {
    "name": "A",
    "easting": 680100,
    "northing": 3975200
}
```

In this dictionary, `"name"`, `"easting"`, and `"northing"` are the **keys**. `"A"`, `680100`, and `680100` are the **values**.
*Values* in a dictionary can be *any* data type.


We retrieve dictionary values using the keys rather than numerical indexes.  This can make code easier to read.


````python
locationDict = {
    "name": "A",
    "easting": 680100,
    "northing": 3975200
}

print(locationDict["name"])
print(locationDict["easting"])
print(locationDict["northing"])
````

## 3. A Dictionary Can Contain a List

A dictionary value can itself be a list.


````python
#redefine locationDict to store coordinates together.
locationDict = {
    "name": "A",
    "coordinates": [680100, 3975200]
}

print(locationDict["coordinates"])
````

Now we can retrieve the list first, and then retrieve an item from that list:

```python
location["coordinates"][0]
```

Read it from left to right:

1. Start with `location`.
2. Retrieve `"coordinates"`.
3. The result is a list.
4. Retrieve index `0` from that list.


````python

print(locationDict["coordinates"][0])
print(locationDict["coordinates"][1])
````

## 5. A GeoJSON Feature

A GeoJSON **Feature** commonly has:

- `properties` — descriptive information;
- `geometry` — spatial information.

For today, we will work with Python dictionaries that follow this GeoJSON structure.


````python
feature = {
    "type": "Feature",
    "properties": {
        "name": "A",
        "category": "Location"
    },
    "geometry": {
        "type": "Point",
        "coordinates": [680100, 3975200]
    }
}
````

### Explore the Feature


````python
print(feature["type"])
print(feature["properties"]["name"])
print(feature["properties"]["category"])
print(feature["geometry"]["type"])
print(feature["geometry"]["coordinates"])
````

## 6. A GeoJSON FeatureCollection

A FeatureCollection contains multiple features.

The value associated with `"features"` is a **list**.

Each item in that list is a **dictionary representing one Feature**.


````python
geojson = {
    "type": "FeatureCollection",
    "features": [
        {
            "type": "Feature",
            "properties": {
                "name": "A",
                "category": "Location"
            },
            "geometry": {
                "type": "Point",
                "coordinates": [680100, 3975200]
            }
        },
        {
            "type": "Feature",
            "properties": {
                "name": "B",
                "category": "Location"
            },
            "geometry": {
                "type": "Point",
                "coordinates": [680450, 3975450]
            }
        },
        {
            "type": "Feature",
            "properties": {
                "name": "C",
                "category": "Location"
            },
            "geometry": {
                "type": "Point",
                "coordinates": [681000, 3976100]
            }
        },
        {
            "type": "Feature",
            "properties": {
                "name": "D",
                "category": "Location"
            },
            "geometry": {
                "type": "Point",
                "coordinates": [680300, 3975650]
            }
        }
    ]
}
````

````python
print(geojson["features"][0]["geometry"]["coordinates"])
````

### FeatureCollection hierarchy

```text
geojson
│
├── type
│
└── features
     │
     ├── Feature 0
     │    ├── properties
     │    │    ├── name
     │    │    └── category
     │    └── geometry
     │         ├── type
     │         └── coordinates
     │
     ├── Feature 1
     ├── Feature 2
     └── Feature 3
```

## 10. Loop Through the Features

Because `geojson["features"]` is a list, we can loop through it.


````python
for feature in geojson["features"]:
    print(feature["properties"]["name"])
````

Inside the loop, `feature` represents one dictionary from the `"features"` list.


````python
for feature in geojson["features"]:
    name = feature["properties"]["name"]
    coordinates = feature["geometry"]["coordinates"]

    print(name, coordinates)
````

## 11. Print a More Useful Sentence


````python
for feature in geojson["features"]:
    name = feature["properties"]["name"]
    coordinates = feature["geometry"]["coordinates"]

    x = coordinates[0]
    y = coordinates[1]

    print(name, "is located at", x, ",", y)
````

## Key idea from today

GeoJSON is hierarchical.

To retrieve information, follow the structure one level at a time, using brackets to go down to the next leve.

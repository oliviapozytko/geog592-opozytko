```python
import pandas as pd
import geopandas as gpd
import matplotlib.pyplot as plt
```

Here, I have created several examples showing how a loop can create labels of different sizes.

```python


# getting the US census data

url_0 = "https://www2.census.gov/geo/tiger/GENZ2024/shp/" # url for census tiger data 
url_county = "cb_2024_us_county_500k.zip" # county level
url_state = "cb_2024_us_state_500k.zip" # state level 


## Code for county level data 

counties = gpd.read_file(url_0+url_county)
nc_counties = counties.loc[counties["STATEFP"] == "37"].copy()
nc_counties.plot(edgecolor="black", facecolor="white")
```

```python
## One loop version

ncc = nc_counties.copy()
ncc["x"] = nc_counties.centroid.x
ncc["y"] = nc_counties.centroid.y

fig, ax = plt.subplots(figsize=(12, 6))

ncc.plot(ax=ax, facecolor="gray", edgecolor="pink")

for name in ncc["NAME"]:
    county = ncc[ncc["NAME"] == name]
    x = county["x"].iloc[0]
    y = county["y"].iloc[0]

    ax.text(x, y, name, fontsize=6, ha="center", va="center")

ax.set_axis_off()
plt.show()
```

```python
ncc = nc_counties.copy()
ncc["x"] = ncc.centroid.x
ncc["y"] = ncc.centroid.y

fig, ax = plt.subplots(figsize=(12, 6))
ncc.plot(ax=ax, facecolor="gray", edgecolor="pink")

for i in ncc.index:
    ax.text(
        ncc.loc[i, "x"],
        ncc.loc[i, "y"],
        ncc.loc[i, "NAME"],
        fontsize=6,
        ha="center",
        va="center"
    )

ax.set_axis_off()
plt.show()
```

```python
ncc = nc_counties.copy()
ncc["x"] = ncc.centroid.x
ncc["y"] = ncc.centroid.y

fig, ax = plt.subplots(figsize=(12, 6))
ncc.plot(ax=ax, facecolor="gray", edgecolor="pink")

for x, y, name in zip(ncc["x"], ncc["y"], ncc["NAME"]):
    ax.text(x, y, name, fontsize=6, ha="center", va="center")

ax.set_axis_off()
plt.show()
```

```python
ncc = nc_counties.copy()
ncc["x"] = ncc.centroid.x
ncc["y"] = ncc.centroid.y

fig, ax = plt.subplots(figsize=(12, 6))
ncc.plot(ax=ax, facecolor="gray", edgecolor="pink")

for index, row in ncc.iterrows():
    ax.text(
        row["x"],
        row["y"],
        row["NAME"],
        fontsize=6,
        ha="center",
        va="center"
    )

ax.set_axis_off()
plt.show()
```

```python
#Most efficient method

fig, ax = plt.subplots(figsize=(12, 6))

nc_counties.plot(ax=ax, facecolor="gray", edgecolor="pink")

for index, row in nc_counties.iterrows():
    center = row.geometry.centroid

    ax.text(
        center.x,
        center.y,
        row["NAME"],
        fontsize=6,
        ha="center",
        va="center"
    )

ax.set_axis_off()
plt.show()
```

```python
## Text based on size of ALAND
fig, ax = plt.subplots(figsize=(12, 6))

nc_counties.plot(ax=ax, facecolor="gray", edgecolor="pink")

for index, row in nc_counties.iterrows():
    center = row.geometry.centroid

    ax.text(
        center.x,
        center.y,
        row["NAME"],
        fontsize=row['ALAND']/100_000_000,
        ha="center",
        va="center"
    )

ax.set_axis_off()
plt.show()
```

```python
# Min-Max Normalization 
w_max = nc_counties["ALAND"].max()
w_min = nc_counties["ALAND"].min()
w_range = w_max - w_min

nc_counties['normalized'] = (nc_counties['ALAND']-w_min) / w_range
max(nc_counties['normalized'])
```

```python
#Most efficient method

fig, ax = plt.subplots(figsize=(12, 6))

nc_counties.plot(ax=ax, facecolor="gray", edgecolor="pink")

for index, row in nc_counties.iterrows():
    center = row.geometry.centroid

    ax.text(
        center.x,
        center.y,
        row["NAME"],
        # fontsize=row['ALAND']/100000000,
        fontsize=(row['normalized']*10)+2,
        ha="center",
        va="center"
    )

ax.set_axis_off()
plt.show()
```
